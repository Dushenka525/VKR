```python
import matplotlib.pyplot as plt
def add_labels(x, y):
    for i in range(len(x)):
        plt.text(i, y[i] / 2, y[i], ha='center')  # Placing text at half the bar height
```

```python
x = ["Линейная регрессия", "Случайные лес", "Градиантный бустинг", "KAN"]
y = [0.0414, 0.0155, 0.0158, 0.0171]
plt.figure(figsize=(10, 5))
# Creating bar chart
plt.bar(x, y)

# Adding value labels
add_labels(x, y)

# Adding title and labels
plt.title("Тестовый набор")
# plt.xlabel("Название модели")
plt.ylabel("MAE")
```
